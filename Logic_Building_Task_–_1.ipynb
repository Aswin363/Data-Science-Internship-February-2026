#1. User Login Check

```python
#Predefined correct credentials
username = "admin"
password = "1234"

#Taking input from user
input_username = input("Enter username: ")
input_password =input("Enter password: ")

#Checking if both username and password match
if input_username == username and input_password == password:
  print("login Successful") # If credentials match
else:
  print("Invalid Credentials") # If credentials do not match
```

#2. Pass / Fail Analyzer


```python
#List of student marks
marks = [45, 78, 90, 33, 60]

#Initialize counters
pass_count = 0
fail_count = 0

#Loop through each mark
for mark in marks:
  if mark >=50:
    pass_count += 1 # Increase pass counter
  else:
    fail_count += 1 # Increase fail counter

# Display final results
print("\nTotal Students:", len(marks))
print("Total Passed:", pass_count)
print("Total Failed:", fail_count)
```

#3. Simple Data Cleaner


```python
#Raw names with extra space and mixed case
names = [" Alice ", "bob", " CHARLIE "]

cleaned_names = []

#Clean each name
for name in names:
  cleaned_name = name.strip().lower()  # Remove space and convert to lowercase
  cleaned_names.append(cleaned_name)   #Add cleaned name to list

  print("\nCleaned Names:", cleaned_names)
```

#4. Message Length Analyzer


```python
messages = ["Hi", "Welcome to the platform", "OK"]

#Check each message
for message in messages:
  message_length = len(message) #Calculate length
  print(f"\nMessage: '{message}'")
  print("Length:", message_length)

#Flag long messages
  if message_length > 10:
    print("This message is longer than 10 characters")
```

#5. Error Message Detector


```python
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

error_count = 0

#Count ERROR entries
for log in logs:
  if log == "ERROR":
    error_count += 1

print("\nTotal ERROR entries:",error_count)
```
