#Problem 1: Mobile Recharge Validation System

```python
# Problem 1: Mobile Recharge Validation System

def validate_recharge():

    valid_plans = [199, 299, 249, 599]

    while True:
        amount = int(input("Enter recharge amount: ₹"))

        # Rule 1: Minimum recharge ₹50
        if amount < 50:
            print("Recharge amount must be at least ₹50. Try again.\n")
            continue

        # Rule 2: Check valid plan
        if amount in valid_plans:
            print("Recharge Successful ")
            break
        else:
            print("Invalid Plan. Available plans:", valid_plans)
            print("Please try again.\n")

# Call function
validate_recharge()
```

#Problem 2: Inventory Reorder Alert System

```python
# Function to check stock levels

def check_inventory(products):

    for item, quantity in products.items():   # Loop through dictionary

        if quantity < 15:
            print(item, "→ Reorder Alert")
        else:
            print(item, "→ Stock OK")


# Sample data
inventory = {
    "Rice": 20,
    "Sugar": 10,
    "Oil": 8,
    "Salt": 25
}

check_inventory(inventory)
```

#Problem 3: Student Result Processing System


```python
# Function to calculate average and result

def check_result(marks):

    total = 0

    # Calculate total marks using loop
    for mark in marks:
        total += mark

    average = total / len(marks)

    print("Average Marks:", average)

    # Check pass or fail
    if average >= 50:
        return "Pass"
    else:
        return "Fail"


student_marks = [60, 45, 70, 55]

print("Result:", check_result(student_marks))
```

#Problem 4: Cab Fare Estimator with Retry Option

```python
# Function to calculate fare

def calculate_fare(distance, peak):

    base_fare = 50
    total_fare = base_fare + (distance * 12)

    # Add 25% extra if peak hour
    if peak.lower() == "yes":
        total_fare = total_fare * 1.25

    return total_fare


# Retry system
while True:

    try:
        km = float(input("Enter distance (km): "))
        peak_hour = input("Is it peak hour? (yes/no): ")

        total_fare = calculate_fare(km, peak_hour)

        print("Total Fare: ₹", total_fare)

        retry = input("Do you want to try again? (yes/no): ")

        if retry.lower() != "yes":
            break

    except ValueError:
        print("Invalid input. Please enter numeric distance.\n")
```

#Problem 5: Employee Attendance Eligibility Checker

```python
# Function to check attendance eligibility

def check_attendance(attendance_list):

    present = 0   # count present days

    # Count how many times "P" appears
    for day in attendance_list:
        if day == "P":
            present += 1

    # Calculate percentage
    percentage = (present / len(attendance_list)) * 100

    # Check eligibility
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# Example
attendance = ["P", "P", "A", "P", "P"]
result = check_attendance(attendance)

print(result)
```

#Problem 6: Password Strength Checker


```python

def validate_password(password):

    number_present = False      # Flag to track digit presence
    special_present = False     # Flag to track special character presence
    allowed_chars = "@#$"       # Accepted special characters

    # Check minimum length requirement
    if len(password) < 8:
        return False

    # Iterate through each character in the password
    for ch in password:

        # Check if character is numeric
        if ch.isdigit():
            number_present = True

        # Check if character is one of the allowed special symbols
        if ch in allowed_chars:
            special_present = True

    # Password is valid only if both conditions are satisfied
    return number_present and special_present


# Retry system (simulating login system)
while True:

    user_input = input("Enter Password: ")

    if validate_password(user_input):
        print("Strong Password")
        break
    else:
        print("Weak Password. Please try again.\n")
```
