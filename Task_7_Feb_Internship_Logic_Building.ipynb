#Problem 1: Smart Light Controller

```python
# Problem 1: Smart Light Controller

class SmartLight:
    def __init__(self, light_name):
        # Store light name and default status
        self.light_name = light_name
        self.status = "OFF"   # Default state

    def turn_on(self):
        self.status = "ON"

    def turn_off(self):
        self.status = "OFF"

    def display_status(self):
        print(f"{self.light_name} is {self.status}")


# ---- Execution ----
light = SmartLight("Bedroom Light")
light.turn_on()
light.display_status()
```

#Problem 2: Employee ID Card System

```python
# Problem 2: Employee ID Card System

class Employee:
    def __init__(self, name, emp_id, department):
        # Initialize employee details
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_id_card(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")


# ---- Execution ----
emp = Employee("Rahul", "EMP102", "IT")
emp.display_id_card()
```

#Problem 3 – Digital Contact Manager

```python
# Mobile Contact Management System

class ContactEntry:

    def __init__(self, person_name, mobile_number):
        self.person_name = person_name
        self.mobile_number = mobile_number

    def save_and_display(self):
        print("\n✔ Contact Successfully Stored")
        print(f"Contact Name  : {self.person_name}")
        print(f"Phone Number  : {self.mobile_number}")


# ---- Execution ----
contact_name = input("Enter Contact Name: ")
phone = input("Enter Phone Number: ")

contact = ContactEntry(contact_name, phone)
contact.save_and_display()
```

#Problem 4: Product Price Tag Generator

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def print_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹" + str(self.price))


# Taking Input
product_name = input("Product Name: ")
product_price = int(input("Price: "))

# Creating Object
item = Product(product_name, product_price)

# Displaying Output
item.print_price_tag()
```

#Problem 5: Movie Rating Display System

```python
class Movie:

    def __init__(self):
        self.title = ""
        self.rating_value = 0.0

    def take_input(self):
        self.title = input("Movie Name: ")
        self.rating_value = float(input("Rating: "))

    def show(self):
        result_movie = "Movie: " + self.title
        result_rating = "Rating: " + str(self.rating_value) + " / 5"
        print(result_movie)
        print(result_rating)


# creating object
m1 = Movie()

# calling functions
m1.take_input()
m1.show()
```

#Problem 6: Delivery Address Manager

```python
class DeliveryInfo:

    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def show_delivery_details(self):
        print("Delivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.address)


# Taking input from user
name = input("Customer Name: ")
addr = input("Address: ")

# Creating object
delivery = DeliveryInfo(name, addr)

# Printing details
delivery.show_delivery_details()
```
