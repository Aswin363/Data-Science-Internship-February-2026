#Problem 1: Social Media – Post Engagement Analyzer

```python
# Social Media Engagement Performance Evaluator

def evaluate_post_performance(engagement_data):

    # Calculate overall engagement using loop
    overall_reactions = 0
    for reaction in engagement_data:
        overall_reactions = overall_reactions + reaction

    # Classify performance level
    if overall_reactions >= 2000:
        performance = "Highly Viral Content"
    elif overall_reactions >= 1000:
        performance = "Trending Post"
    else:
        performance = "Standard Reach"

    return overall_reactions, performance


# Sample Input Data
reaction_list = [200, 300, 150, 400]

likes_total, post_category = evaluate_post_performance(reaction_list)

print(f"Engagement Count → {likes_total}")
print(f"Performance Level → {post_category}")
```

#Problem 2: Healthcare – Medicine Stock Alert System

```python
# Healthcare Inventory Monitoring System

def medicine_inventory_checker(current_stock):

    # Determine stock condition
    if current_stock <= 0:
        stock_status = "Out of Stock - Immediate Restock Required"

    elif current_stock < 10:
        stock_status = "Low Inventory Warning"

    else:
        stock_status = "Inventory Level Stable"

    return stock_status


# Sample Input
available_units = 6

result_status = medicine_inventory_checker(available_units)

print(f"Available Medicine Units → {available_units}")
print(f"Inventory Status → {result_status}")
```

#Problem 3 – Agriculture: Rainfall Adequacy Checker

```python
# Climate Data Evaluation Module

def rainfall_decision_system(season_data, benchmark):

    # Using sum() directly instead of loop
    mean_value = round(sum(season_data) / len(season_data), 2)

    decision_map = {
        True: "Rainfall Requirement Achieved",
        False: "Rainfall Requirement Not Achieved"
    }

    return mean_value, decision_map[mean_value >= benchmark]


# Example Data
season_readings = [70, 75, 68, 80, 67]
required_limit = 70

average_value, rainfall_result = rainfall_decision_system(season_readings, required_limit)

print("Computed Average:", average_value)
print("Decision:", rainfall_result)
```

#Problem 4: Social Media – Duplicate Account Detection

```python
# Function to check whether usernames are unique or duplicated
def verify_unique_accounts(user_list):

    # Convert list into set to remove duplicate values
    unique_users = set(user_list)

    # If length differs, duplicates exist
    if len(unique_users) != len(user_list):
        return "Duplicate Profiles Detected"
    else:
        return "All Accounts Unique"


# Sample username list
registered_users = ["aswin", "anita", "aswin", "kiran"]

# Display result
print(f"Account Verification Result → {verify_unique_accounts(registered_users)}")
```

#Problem 6: Agriculture – Premium Crop Pricing Filter

```python
def identify_premium_crops(price_collection):

    premium_list = []

    for price in price_collection:
        if price > 2000:
            premium_list.append(price)

    return premium_list


crop_prices = [1500, 2500, 1800, 3200]

premium_items = identify_premium_crops(crop_prices)

print(f"Premium Crop Prices → {premium_items}")
```

#Problem 7: IT Monitoring – Application Stability *Analyzer*

```python
# Function to analyze system condition based on total errors
def analyze_system_condition(total_errors):

    # Default system state
    system_state = ""

    # Check error levels
    if total_errors == 0:
        system_state = "Application Running Smoothly"

    elif 1 <= total_errors <= 5:
        system_state = "Minor Performance Warnings"

    else:
        system_state = "High Risk - Immediate Attention Required"

    return system_state


# Sample monitoring data
logged_errors = 7

print(f"Total Errors Detected → {logged_errors}")
print(f"Current System Condition → {analyze_system_condition(logged_errors)}")
```

# Problem 8: Banking – Transaction Authorization System

```python
def authorize_transaction(transaction_amount):

    daily_limit = 50000

    if transaction_amount <= daily_limit:
        decision = "Transaction Approved"
    else:
        decision = "Transaction Declined (Limit Exceeded)"

    return decision


amount = 60000

print(f"Requested Amount → ₹{amount}")
print(f"Authorization Status → {authorize_transaction(amount)}")
```

# Problem 9: E-Learning – Attendance Compliance Checker

```python
# Student Attendance Evaluation System

def check_student_exam_status(records):

    # Ensure attendance list is not empty
    if not records:
        return 0, "No Attendance Data"

    # Calculate total presence using built-in sum
    total_present = sum(records)
    total_classes = len(records)

    # Compute attendance ratio
    attendance_ratio = total_present / total_classes
    attendance_percent = round(attendance_ratio * 100, 2)

    # Determine exam permission using conditional expression
    exam_permission = (
        "Eligible to Appear"
        if attendance_percent >= 75
        else "Shortage of Attendance"
    )

    return attendance_percent, exam_permission


# Monitoring data
attendance_log = [1, 1, 1, 0, 1, 1, 1, 1, 0, 1]

percentage_result, permission_result = check_student_exam_status(attendance_log)

print("Attendance Computed:", percentage_result, "%")
print("Exam Status:", permission_result)
```

#Problem 10: *Smart* Electricity Billing Engine

```python

# Smart Residential Electricity Billing System
# This function calculates electricity charges using
# dynamic slab processing and classifies consumption level.


def process_electricity_usage(total_units):

    # Define slab configuration (limit, rate)
    pricing_slabs = [
        (100, 3),             # First 100 units
        (100, 5),             # Next 100 units
        (float("inf"), 7)     # Units above 200
    ]

    remaining_units = total_units
    final_bill = 0

    # Calculate bill dynamically for each slab
    for unit_limit, price_per_unit in pricing_slabs:

        if remaining_units <= 0:
            break

        units_to_charge = min(remaining_units, unit_limit)
        final_bill += units_to_charge * price_per_unit
        remaining_units -= units_to_charge


    # Determine usage category
    if final_bill < 500:
        consumption_type = "Low Energy Consumption"
    elif final_bill <= 1500:
        consumption_type = "Moderate Energy Consumption"
    else:
        consumption_type = "High Energy Consumption"


    return final_bill, consumption_type


# Example input
units_consumed = 250

bill_amount, usage_level = process_electricity_usage(units_consumed)

print(f"Units Consumed → {units_consumed}")
print(f"Total Bill Amount → ₹{bill_amount}")
print(f"Consumption Category → {usage_level}")
```
