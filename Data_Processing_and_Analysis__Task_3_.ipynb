#Problem 1: Employee Performance Bonus Eligibility

```python
# Problem 1: Employee Performance Bonus Eligibility

employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}


#First, I find the highest score using max().

highest_score = max(employees.values())

# Then I check which employees have that highest score (to handle tie cases).

top_employees = []

for name, score in employees.items():
  if score == highest_score:
    top_employees.append(name)

    # Finally, I print the eligible employees.
    print("Top Performers Eligible for Bonus:", ", ".join(top_employees), f"(Score: {highest_score})")


```

#Problem 2 – Search Keyword Analysis


```python
# Problem 2: Search Query Keyword Analysis

# I am taking the raw search input entered by a user
search_input = "Buy mobile phone buy phone online"

# Step 1: Normalize text (convert to lowercase and remove punctuation)
import string
normalized_text = search_input.lower().translate(
    str.maketrans("", "", string.punctuation)
)

# Step 2: Break the sentence into individual words
word_list = normalized_text.split()

# Step 3: Build a frequency dictionary
frequency_map = {}

for item in word_list:
    frequency_map[item] = frequency_map.get(item, 0) + 1

# Step 4: Display only repeated keywords
print("Repeated Keywords:")
for key, value in frequency_map.items():
    if value > 1:
        print(f"{key} -> {value}")
```

#Problem 3 – Sensor Data Validation

```python
# Problem 3: Sensor Data Validation

# Raw sensor readings collected hourly
sensor_data_list = [3, 4, 7, 8, 10, 12, 5]

# Extract only valid (even) readings along with their hour index
validated_output = [
    (index, reading)
    for index, reading in enumerate(sensor_data_list)
    if reading % 2 == 0
]

print("Valid Sensor Reading Report")
print("Format: (Hour Index, Reading Value)")
print(validated_output)
```

#Problem 4 – Email Domain Analysis

```python
# Problem 4: Email Domain Usage Analysis

# List of employee email addresses
employee_email_list = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

# Step 1: Extract only domain names
extracted_domains = [mail.split("@")[-1] for mail in employee_email_list]

# Step 2: Count frequency of each domain
domain_frequency = {}

for domain_name in extracted_domains:
    domain_frequency[domain_name] = domain_frequency.get(domain_name, 0) + 1

# Step 3: Calculate percentage usage
total_users = len(employee_email_list)

print("Email Domain Usage Report")

for domain_name, freq in domain_frequency.items():
    usage_percent = (freq / total_users) * 100
    print(f"{domain_name:<12} → {freq} users ({usage_percent:.0f}%)")
```

# Problem 5: Sales Spike Detection


```python
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Step 1: Calculate average
average = sum(sales) / len(sales)

# Step 2: Calculate threshold (30% above average)
threshold = average * 1.3

# Step 3: Detect spikes
for i in range(len(sales)):
    if sales[i] > threshold:
        print(f"Day {i+1}: {sales[i]}")
```

# Problem 6: Duplicate User ID Detection

```python
# Problem 6: Duplicate User ID Detection

# Input list of user IDs
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Step 1: Create an empty dictionary to store counts
id_count = {}

# Step 2: Count occurrences of each user ID
for user in user_ids:
    if user in id_count:
        id_count[user] += 1
    else:
        id_count[user] = 1

# Display user IDs that appear more than once in the system
print("Duplicate User IDs:")
for user, count in id_count.items():
    if count > 1:
        print(f"{user} → {count} times")
```
